# Prueba de rachas

Consiste en comparar cada número con el anterior. Si el nuevo número es mayor, anotamos un + (sube). Si es menor, anotamos un - (baja). Luego contamos cuántos bloques (o rachas) de signos iguales se formaron.

Las hipotesis son:
  $$H_0: \text{Los numeros de la muestra } \{x_i\} \text{ son independientes}$$
  $$H_1: \text{Los numeros de la muestra } \{x_i\} \text{ no son independientes}$$

Las fórmulas matemáticas para una muestra de tamaño $N$ son:
$$\mu_{C_0} = \frac{2n-1}{3}$$
$$\sigma^2_{C_0} = \frac{16n-29}{90}$$
$$Z_0 = \frac{C_0 - \mu_{C_0}}{\sigma_{C_0}}$$

Donde $Z_0$ es el estadistico de prueba

```python
import numpy as np
import scipy.stats as st
```

```python
def prueba_corridas_arriba_abajo(datos, alfa=0.05):
    print("--- PRUEBA DE RACHAS ---")
    N = len(datos)

    signos = []   #Genera los signos
    for i in range(1, N):
        if datos[i] > datos[i-1]:
            signos.append('+')
        else:
            signos.append('-')

    R = 1    #Cuenta las rachas
    for i in range(1, len(signos)):
        if signos[i] != signos[i-1]:  # Si el signo actual es diferente al anterior, es una nueva corrida
            R += 1

    mu_R = (2 * N - 1) / 3   #Media
    sigma_R = np.sqrt((16 * N - 29) / 90)   #Desviación

    Z_calc = (R - mu_R) / sigma_R    #Zo

    Z_critico = st.norm.ppf(1 - alfa/2)  #Valor critico con prueba de 2 colas

    print(f"Total de datos: {N}")
    print(f"Número de rachas observadas: {R}")
    print(f"Media esperada: {mu_R:.4f}")
    print(f"Desviación estándar: {sigma_R:.4f}")
    print(f"Estadístico Z calculado: {Z_calc:.4f}")
    print(f"Valor crítico Z (alfa={alfa}): {Z_critico:.4f}\n")

    if abs(Z_calc) < Z_critico:
        print("Conclusión: |Z calculado| < Z crítico")
        print("NO se rechaza H_0. Los números son independientes).")
    else:
        print("Conclusión: |Z calculado| >= Z crítico")
        print("SE RECHAZA H_0. Los números NO son independientes.")
```

```python
prueba_corridas_arriba_abajo(data, alfa=0.05)
```
